```python
# Complete preprocessing workflow for the analyzed dataset
import pandas as pd

df = pd.read_csv("/Users/harshit9793/ML plateform/50_Startups.csv")
print(f"Loaded dataset: {df.shape[0]} rows × {df.shape[1]} columns")

# Apply preprocessing only to the selected column
df["R&D Spend"] = pd.to_numeric(df["R&D Spend"], errors="coerce")
df["R&D Spend"] = df["R&D Spend"].fillna(df["R&D Spend"].median())

# Check the result
print(f"Dataset shape after preprocessing: {df.shape}")
print("Missing values remaining in " + "R&D Spend" + ": " + str(df["R&D Spend"].isna().sum()) if "R&D Spend" in df.columns else "Selected column was dropped.")

# Preview only the selected column or its encoded output columns
print(df[["R&D Spend"]].head())
```
